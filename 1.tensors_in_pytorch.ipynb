## Mathematical Operations

```python
x = torch.rand(2,2)
print(x)
```

```python
# addition
x + 2
# subtraction
x - 2
# multiplicatoin
x * 2
# division
x / 2
# int division
(x * 100)//3
# mod
((x * 100)//3)%2
# pow
x ** 2
```

```python
a = torch.rand(2,2)
b = torch.rand(2,2)

print(a)
print(b)
```

```python
# add
a + b
# sub
a - b
# multiply
a * b
# division
a / b
# power
a ** b
# mod
a % b
```

```python
# abs
c = torch.tensor([1, -2, 3, -4])
torch.abs(c)
```

```python
# negative
torch.neg(c)
```

```python
d = torch.tensor([1.9, 2.3, 3.7, 4.4])
```

```python
# round
torch.round(d)
```

```python
torch.ceil(d)
```

```python
torch.floor(d)
```

```python
torch.clamp(d, min=2, max=3)
```

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
# total sum
torch.sum(e)
# column wise sum
torch.sum(e, dim=0)
# row wise sum
torch.sum(e, dim=1)
```

```python
# mean
torch.mean(e)
```

```python
# median
torch.median(e)
```

```python
# max and min
torch.max(e), torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax and argmin
torch.argmax(e), torch.argmin(e)
```
