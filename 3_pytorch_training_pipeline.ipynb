```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
df.head()
```

**TRAIN TEST SPLIT**

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:,0], test_size=0.2)
```

**STANDARD SCALING: Normalizing all the values in a given range**

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
print(y_train)
```

**LABEL ENCODING: Converting all the characters to numbers like (M,B) = (1,0)**

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.fit_transform(y_test)
```

```python
print(y_train)
```

```python
X_train_tensor = torch.tensor(X_train)
X_test_tensor = torch.tensor(X_test)
y_train_tensor = torch.tensor(y_train)
y_test_tensor = torch.tensor(y_test)
```

```python
X_train_tensor.shape
```

```python
y_train_tensor.shape
```

**Defining the Model**

```python
class SimpleNN():
    def __init__(self, X):
        self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
        self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

    def forward(self, X):
        z = torch.matmul(X, self.weights) + self.bias
    #   z = X @ self.weights + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss_function(self, y_pred, y):
        epsilon = 1e-7 # A very very small number to avoid problem caused by zero itself as log(0) = -inf

        y_pred = torch.clamp(y_pred, epsilon, 1-epsilon)
        # torch.clamp(tensor, min, max) => Any value less than min make it min, if greater than max make it max.

        # Calculate loss
        loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1-y_pred)).mean()

        return loss
```

```python
learning_rate = 0.1
epochs = 101
```

**Training Pipeline**

```python
model = SimpleNN(X_train_tensor)

for epoch in range(epochs):

    # forward pass
    y_pred = model.forward(X_train_tensor)

    # loss calculate
    loss = model.loss_function(y_pred, y_train_tensor)

    # backward pass
    loss.backward()

    # update parameters
    with torch.no_grad():
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

    # zero gradient
    model.weights.grad.zero_()
    model.bias.grad.zero_()

    # print loss after every 10 epoch
    if epoch%10 == 0:
        print(f"Epoch: {epoch} | Loss: {loss.item()}")
```

```python
# model evaluation

with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred = (y_pred > 0.9).float()
    accuracy = (y_pred == y_test_tensor).float().mean()
    print(f"Accuracy: {accuracy.item()}")
```
